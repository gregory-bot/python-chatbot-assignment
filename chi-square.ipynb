```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
#load the data 
sb_data=pd.read_csv(r"superbowl.csv")
sb_data.tail()
```

```python
# Check the counts of each coin toss winner
sb_data["Coin Toss Winner"].value_counts()
```

```python
# Check the counts of each game winner
sb_data["Game Winner"].value_counts()
```

```python
#based on the coin toss winner we can say its not a fair coin
```

```python
coin_toss_counts = sb_data["Coin Toss Winner"].value_counts().values
```

```python
fair_coin_expected=[sum(coin_toss_counts)/2,sum(coin_toss_counts)/2]
fair_coin_expected
```

```python
fair_coin_observed = coin_toss_counts
fair_coin_observed
```

```python
fair_coin_results=stats.chisquare(fair_coin_observed,fair_coin_expected)
fair_coin_results
```

```python
game_winner_counts = sb_data["Game Winner"].value_counts().values
game_winner_counts
```

```python
game_winner_expected=[sum(game_winner_counts)/2,sum(game_winner_counts)/2]
game_winner_expected
```

```python
game_winner_observed = game_winner_counts
game_winner_observed
```

```python
game_winner_results=stats.chisquare(game_winner_observed,game_winner_expected)
game_winner_results
```

```python
nfl_games=pd.read_csv(r"nfl_games.csv")
nfl_games.head()
```

```python
# Narrow to relevant rows
nfl_games = nfl_games[(nfl_games["neutral"] == 0) & (nfl_games["playoff"] == 0) & (nfl_games["season"] >= 1966 )]
# Engineer a new feature
nfl_games.loc[nfl_games["result1"] == 0, "Game Winner"] = "Away Team"
nfl_games.loc[nfl_games["result1"] == 1, "Game Winner"] = "Home Team"

game_winner_counts_nfl = nfl_games["Game Winner"].value_counts().sort_index()
game_winner_counts_nfl
```

```python
nfl_expected = [sum(game_winner_counts_nfl)/2, sum(game_winner_counts_nfl)/2]
nfl_expected
```

```python
nfl_chisquare=stats.chisquare(game_winner_counts_nfl, nfl_expected)
nfl_chisquare
#based on this how do we conclude 
#p-value is less than 0.05 so we reject the null hypothesis
#and thus hometeam have advantage
```

```python
#is winning the game relatted to winning the coin toss
#what is the hypothesis here
#null hypothesis: there is no relation between winning the coin toss and winning the game
#alternate hypothesis: there is a relation between winning the coin toss and winning the game
```

```python
independence_table = pd.crosstab(sb_data["Coin Toss Winner"], sb_data["Game Winner"])
independence_table
```

```python
chi2, p, dof, expected = stats.chi2_contingency(independence_table)
print(chi2, p, dof, expected)
```
