```python
import pandas as pd
import numpy as np

# Read the file without assuming the first row is header
df = pd.read_excel('loan_limit_increases.xlsx', header=None)

print("Raw data (first 10 rows):")
print(df.head(10))

# The column names are in row 0 (the first row of data)
# The actual data starts from row 1
column_names = df.iloc[0].tolist()

print("\nColumn names from first row:")
print(column_names)

# Create a new dataframe starting from row 1 with proper column names
df_clean = df.iloc[1:].copy()
df_clean.columns = column_names

# Reset index
df_clean.reset_index(drop=True, inplace=True)

print("\nCleaned dataframe:")
print("Columns:", df_clean.columns.tolist())
print("Shape:", df_clean.shape)
print("\nFirst 3 rows:")
print(df_clean.head(3))
```

```python
import pandas as pd
import numpy as np

# Read the file without any assumptions
df_raw = pd.read_excel('loan_limit_increases.xlsx', header=None)

print("=== DEBUGGING STEP ===")
print("First 5 rows of raw data:")
for i in range(5):
    print(f"Row {i}: {list(df_raw.iloc[i])}")

# Based on your earlier output, let's manually extract the data
# Row 0 seems to be: ['loan_limit_increases', 'Unnamed: 1', 'Unnamed: 2', ...]
# Row 1 seems to be the actual column names: ['Customer ID', 'Initial Loan ($)', ...]

# Let's manually create the column names and data
if len(df_raw) > 1:
    # Get the column names from row 1
    column_names = df_raw.iloc[1].tolist()
    print(f"\nColumn names from row 1: {column_names}")
    
    # Get the data starting from row 2
    data_rows = []
    for i in range(2, len(df_raw)):
        data_rows.append(df_raw.iloc[i].tolist())
    
    print(f"\nFirst 3 data rows:")
    for i in range(min(3, len(data_rows))):
        print(f"Data row {i}: {data_rows[i]}")
    
    # Create the DataFrame
    df = pd.DataFrame(data_rows, columns=column_names)
    
    print(f"\n✅ Created DataFrame with shape: {df.shape}")
    print("DataFrame columns:", df.columns.tolist())
    print("\nFirst 3 rows of cleaned DataFrame:")
    print(df.head(3))
else:
    print("Not enough rows in the file!")
```
